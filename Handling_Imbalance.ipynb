### 1. Import Necessary libraries

```python
from sklearn import datasets
import pandas as pd
```

### 2. Import dataset

```python
breast_cancer_data = datasets.load_breast_cancer()
breast_cancer_data
print(breast_cancer_data.DESCR)
```

```python
pd.DataFrame(breast_cancer_data.data) # Data key has the input values and target has the ouput values ,
# feature name is the name of the columns
```

## 2.1 Assigning the column values

```python
pd.DataFrame(data=breast_cancer_data.data, columns=breast_cancer_data.feature_names)
```

## 2.2 Adding the Target column to the dataset

```python
breast_cancer_data_df =  pd.DataFrame(data=breast_cancer_data.data, columns=breast_cancer_data.feature_names)
breast_cancer_data_df["Target"] = breast_cancer_data.target
breast_cancer_data_df
```

### 3. Data Understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

### 4. Model Building

```python
X=breast_cancer_data_df.drop(labels = "Target", axis = 1 )
y=breast_cancer_data_df["Target"]
```

```python
X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=1234,shuffle=True,stratify=y)
```

```python
X_train.shape , y_train.shape , X_test.shape, y_test.shape
```

#### Handling the Imbalance 

To Balance the % percentage of data in Training and testing as like in Actual Data
----------------------------------------------------------------------------------
1. Finding the data split in orginal / Train / Test data
2. Changing the Stratify = y (because the output split is in this column, so we are giving the y value(Target)) in Train Test Split method
   to balance the data percentage among the train and test datas.

To Improve the accuracy of the model predictions we are telling to give more weightage to the class which we are trying to predict
----------------------------------------------------------------------------------------------------------------------------------
1. Tweaking the hyper parameter value, Class Weight in dt_model method (here we used decision tree model)
2. Class Weight in Decision tree clasifier {class_label: weight} - here our class labels are (0- denotes the Malignant and 1- denotes the Benign), since data is in ration of 37:63 giving the class weightage as (0:2, 1:1)

positive prediction  - What is the exact class i am intrested to predict (here im intrested to predict about the malignant cancer, means im preparing the model the predict this cancer accurately)
Negative prediction  - least bother about (Benign cancer)
   

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt
plt.pie(y.value_counts(), labels = ["Benign Cancer","Malignant Cancer"], autopct = "%1.0f%%" )
plt.show()
```

```python
plt.pie(y_train.value_counts(), labels = ["Benign Cancer","Malignant Cancer"], autopct = "%1.0f%%" )
plt.show()
```

```python
plt.pie(y_test.value_counts(), labels = ["Benign Cancer","Malignant Cancer"], autopct = "%1.0f%%" )
plt.show()
```

### 5.Model Training

```python
%%time
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=3, class_weight = {0: 2, 1:1})
dt_model.fit(X_train, y_train)
```

### 6. Model Testing and 7. Model Evaluation

### Training data

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
print(classification_report(y_train, y_pred_train))
```

```python
print(confusion_matrix(y_train, y_pred_train))
```

### Test data 

```python
y_pred_test = dt_model.predict(X_test)
y_pred_test
```

```python
print(classification_report(y_test, y_pred_test))
```

```python
print(confusion_matrix(y_test, y_pred_test))
```
